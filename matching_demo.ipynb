# Demo Matching 🎯

Follow these simple steps to find your best match:

1. **Select your topic of interest**  
2. **Answer a few questions**  
3. **Find your best match! 🎉**

```python
# Import libraries
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import ast
from utils import *

# Set seed
np.random.seed(42)
```

# Load and Prepare Dataset

```python
# Load WWS dataset
df_clean = pd.read_csv('data/df_clean.csv')

# Load all relevant questions
df_meta_answers = pd.read_excel('data/hand_written_answers.xlsx')
# Convert dict that were saved as str
df_meta_answers['possible_answers'] = df_meta_answers.possible_answers.apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) and x.strip().startswith('{') else x
)

# Select all topics
possible_topics = df_meta_answers['topic'].unique()

# Remove Demographic (not used for matching)
possible_topics = possible_topics[:-1]

# Add Global Survey Mode
possible_topics = np.append(possible_topics, 'Global')

print(f"There is a selection of {df_meta_answers.shape[0]} questions across {len(possible_topics)} topics")
df_meta_answers.head()
```

# Select your topic

_Note: Mode Global englobes all categories_

```python
# TOPIC CHOICE
print("Which topic do you pick?")
for i, topic in enumerate(possible_topics, 1):
    print(f"{i}. {topic}")

choice = int(input("Enter the number of your choice: "))
selected_topic = possible_topics[choice - 1]

print(f"You picked: {selected_topic}")
```

```python
# Select N questions of the selected topic and remove participant who did not answer
selected_questions, df_valid_answers = select_random_questions_and_clean(
    df_meta_answers, df_clean, selected_topic=selected_topic, nbr_of_questions=4
)
```

# 3. Find your best match
After each question you answer, you'll see your top 5 matching countries and your updated compatibility score. At the end, your best matching country will be highlighted, and all scores will be visualized on a global heatmap.

```python
user_answers, top_matches = run_matching_quiz(selected_questions, df_valid_answers)
print(f"Your final match is {top_matches.B_COUNTRY_ALPHA.iloc[0]} with a score of {round(top_matches['Score (%)'].iloc[0])}% 🎉")
```
