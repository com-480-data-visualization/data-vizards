```python
def compute_distribution(data, group_col, selected_question, group_labels=None):
    grouped = data.groupby([group_col, selected_question]).size().reset_index(name='count')
    grouped['percent'] = grouped.groupby(group_col)['count'].transform(lambda x: (x / x.sum()) * 100).round(2)

    result = []
    for group_value, group_data in grouped.groupby(group_col):
        label = group_labels.get(group_value, str(group_value)) if group_labels else str(group_value)
        group_dict = {
            'attribute': label,
            'values': []
        }
        for _, row in group_data.iterrows():
            group_dict['values'].append({
                'range': str(row[selected_question]),
                'value': row['percent']
            })
        result.append(group_dict)
    
    return result


# Setup
selected_question = 'Q29'
selected_country = 'AND'
Gender_idx = 'Q260'
Age_idx = 'Q287'
Religion_idx = 'Q289'
Ethnic_idx = 'Q290'

# Filter
mask = (data_clean['B_COUNTRY_ALPHA'] == selected_country) & \
       (data_clean[[selected_question, Gender_idx, Age_idx, Religion_idx, Ethnic_idx]] > 0).all(axis=1)
data_clean_country = data_clean[mask]


# Labels
gender_labels = {1: 'Male', 2: 'Female'}
age_labels = {
    1: '15-24', 2: '25-34', 3: '35-44', 4: '45-54',
    5: '55-64', 6: '65+'
}
religion_labels = {
    0: 'No',
    1: 'Roman Catholic',
    2: 'Protestant',
    3: 'Orthodox',
    4: 'Jew',
    5: 'Muslim',
    6: 'Hindu',
    7: 'Buddhist',
    8: 'Other (text)'
}

if data_clean_country[selected_question].unique().shape[0] == 10:
    data_clean_country[selected_question].map(map_range)

# Compute distributions
gender_distribution = compute_distribution(data_clean_country, Gender_idx, selected_question, gender_labels)
age_distribution = compute_distribution(data_clean_country, Age_idx, selected_question, age_labels)
religion_distribution = compute_distribution(data_clean_country, Religion_idx, selected_question, religion_labels)

# Final data to compute all histograms
dict_data = {
    "Gender": gender_distribution,
    "Age": age_distribution,
    "Religion": religion_distribution
}

# Display
import pprint
print("GENDER DISTRIBUTION:")
pprint.pprint(gender_distribution)
print("\nAGE DISTRIBUTION:")
pprint.pprint(age_distribution)
print("\nRELIGION DISTRIBUTION:")
pprint.pprint(religion_distribution)
```
