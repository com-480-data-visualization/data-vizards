# Ethnicity mapping
In this notebook we extract the column corresponding to Q290 (with ethnic group indices varying for each country) and add the following information:

- 1: 'White', 
- 2: 'Black',
- 3: 'South Asian',
- 4: 'East Asian',
- 5: 'Arabic/Central Asian',
- 6: 'Southeast Asian',
- 7: 'Indigenous',
- 8: 'Other'

As well as the explicit name given to the ethnic group and the 3-letter country code.

```python
import pandas as pd

# Read only the Q290 column from the large CSV
df = pd.read_csv('../../data/DF_Clean.csv', usecols=['Q290'])

# Save it to a new CSV file
df.to_csv('../../data/Q290.csv', index=False)
```

```python
# Load the Q290 column
q290 = pd.read_csv('../../data/Q290.csv')

# Load the mapping
mapping = pd.read_csv('../../data/ethnicity_map.csv', usecols=['code', 'mapped_category', 'ethnicity_string', 'country'])
```

```python
mapping
```

```python
# For each row in Q290, find the corresponding mapped_category in the mapping
q290['Q290'] = q290['Q290'].astype(str)
mapping['code'] = mapping['code'].astype(str)
mapping['ethnicity_string'] = mapping['ethnicity_string'].astype(str)
mapping['country'] = mapping['country'].astype(str)

# Map Q290 codes to mapped_category
q290['mapped_category'] = q290['Q290'].map(dict(zip(mapping['code'], mapping['mapped_category'])))

# Map Q290 codes to ethnicity strings
q290['ethnicity_string'] = q290['Q290'].map(dict(zip(mapping['code'], mapping['ethnicity_string'])))

# Map Q290 codes to 3-letter country code 
q290['country'] = q290['Q290'].map(dict(zip(mapping['code'], mapping['country'])))

# Save the result as a single-column CSV (header: Q290)
q290.to_csv('../../data/Q290_with_category.csv', index=False)

```
